## 12. Tips & Common Pitfalls
* Always check `value_counts()` **before and after** every fix.
* Don't impute blindly: a hospital stay for sickle-cell patients rarely involves a 70-year-old — use **group-wise** medians.
* Keep a copy of the raw file; never overwrite it.
* Avoid hard-coding: build mapping dictionaries so your steps can be re-run.
* In Power BI, set **Summarize by → Don't summarize** for Age and Year so they are not summed.
* Test slicers: every visual should respond sensibly.
